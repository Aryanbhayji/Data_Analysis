```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Titanic-Dataset.csv").dropna()
df.head(10)
```

```python
#What was the average age of Titanic passengers?

avg_age=df["Age"].mean()
print(avg_age)
```

```python
#What is the middle age of passengers?

middle_age=df["Age"].median()
print(middle_age)
```

```python
#How much does passenger age vary from the average age?

variance=df["Age"].var()
print(variance)
```

```python
#What is the probability that a randomly selected passenger survived?

survival_probability = df['Survived'].mean()

print(survival_probability)


female_survival = df[df['Sex'] == 'female']

probability = female_survival['Survived'].mean()

print(probability)
```

```python
#Is there a relationship between passenger age and ticket fare?

correlation = df['Age'].corr(df['Fare'])

print(correlation)
```

```python
#What percentage of passengers survived?

survival_percentage = (
    df['Survived'].value_counts(normalize=True) * 100
)

print(survival_percentage)
```
