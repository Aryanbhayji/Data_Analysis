```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Titanic-Dataset.csv").dropna()
df.head(10)
```

```python
#Survival rate by gender

sns.barplot(df,x="Survived",hue="Sex")
plt.title("Survival rate by gender")
plt.tight_layout()
```

```python
#Survival rate by passenger class

sns.barplot(df,x="Survived",hue="Pclass")
plt.title("Survival rate by passenger class")
plt.tight_layout()
```

```python
#Relationship between age and survival

subset=df[["Age","Survived"]].corr()

sns.heatmap(subset,annot=True)

```

```python
#Relationship between ticket fare and survival


subset_2=df[["Fare","Survived"]].corr()

sns.heatmap(subset_2,annot=True)

```
