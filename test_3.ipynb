```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data= pd.read_csv("SampleSuperstore.csv").dropna()
data.head(10)
```

```python
#what are the top 10 most profitable product?

top_10_profitable_product= data.groupby("Sub-Category", as_index=False)["Profit"].sum().sort_values(by="Profit",ascending=False)
product=top_10_profitable_product.head(10)

sns.barplot(data=product.sort_values(by="Profit",ascending=False),y="Sub-Category",x="Profit")
print(f'top_10_profitable_product :{product.sort_values(by="Profit",ascending=False,ignore_index=True)}')

```

```python
df_2=pd.read_excel("Online Retail.xlsx")
df_2.head(10)
```

```python
df_2.shape
df_2.info()
```

```python
df_2.describe()
```

```python
subset= df_2[["Quantity","UnitPrice"]]
subset.head(10)
```

```python
#total revenue

df_2["revenue"]=df_2["Quantity"]*df_2["UnitPrice"]
total_revenue=df_2["revenue"].sum()
print(total_revenue)
```

```python
df_2.nunique()
```

```python
#Top_10_Country_Revenue

sub=df_2.groupby("Country",as_index=False)["revenue"].sum()
plt.figure(figsize=(12,7))
top_10=sub.head(10).sort_values(by="revenue",ascending=False)
sns.barplot(data=top_10,x="Country",y="revenue")
plt.title("Top_10_Country_Revenue")
```

```python
#distribution of unit price

prices = df_2.loc[df_2["UnitPrice"] > 0, "UnitPrice"].dropna()
plt.figure(figsize=(10, 5))

sns.histplot(
    prices,
    bins=50,
    kde=True
)

plt.title("Distribution of Unit Prices")
plt.xlabel("Unit Price ")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
