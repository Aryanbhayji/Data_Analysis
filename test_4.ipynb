```python
import seaborn as sns 
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv("winequality-red.csv",sep=";").dropna()
data.head(10)
```

```python
print(data.dtypes)
```

```python

# Calculate correlation between numerical columns
correlation = data.corr(numeric_only=True)

# Create the heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Wine Quality Correlation Heatmap")
plt.show()
```
