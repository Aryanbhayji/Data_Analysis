```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data= pd.read_csv("SampleSuperstore.csv").dropna()
data.head(10)
```

```python
#which category generates the lowest profit                               

lowest_profit= data.groupby("Sub-Category", as_index=False)["Profit"].sum()
print(lowest_profit)


sns.barplot(data=lowest_profit,x="Profit",y="Sub-Category")
print(f'lowest_profit_sub-category: {'Fasteners'}')
print(f'lowest_profit:{"949.5182"}')
plt.show()
```

```python
#checking correlation between discount and profit

Relationship=data.groupby("Discount",as_index=False)["Profit"].sum()
print(Relationship)

sns.scatterplot(data=Relationship,x="Profit",y="Discount")
sns.regplot(data=Relationship,x="Profit",y="Discount")

correlation=data["Discount"].corr(data["Profit"])
print("correlation between Discount and Profit:",correlation)
```
