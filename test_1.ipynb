```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data= pd.read_csv("SampleSuperstore.csv").dropna()
data.head(10)
```

```python
subset=data[["Region","Profit","Category"]]
print(subset.head(10))
```

```python
sns.barplot(x='Region', y='Profit', data=subset,hue='Category', alpha=0.5)
```

```python
subset_2=data[["Region","Category",'Quantity']]
print(subset_2.head(10))
```

```python
sns.kdeplot(x='Quantity', data=subset_2, hue='Category', fill=True, alpha=0.5)
```

```python
subset_3=data[["City","Sales"]]
print(subset_3.head(10))
```

```python
#watching the top 15 cities by technology sales

technology_by_city = (
    data[data["Category"] == "Technology"].groupby("City", as_index=False)["Sales"].sum().sort_values("Sales", ascending=False)
)

display(technology_by_city.head(15))

top_cities = technology_by_city.head(15)

plt.figure(figsize=(10, 6))
sns.barplot(data=top_cities, x="Sales", y="City", color="steelblue")
plt.title("Top 15 Cities by Technology Sales")
plt.xlabel("Total sales ($)")
plt.ylabel("City")
plt.tight_layout()
plt.show()
```
