```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data= pd.read_csv("SampleSuperstore.csv").dropna()
data.head(10)
```

```python
sales_revenue=data['Sales'].sum()
total_profit=data['Profit'].sum()
print("Total Profit: ", total_profit)
print("Total Sales Revenue: ", sales_revenue)
investment=data['Sales'].sum()-data['Profit'].sum()
print("Total Investment: ", investment)

avg_sale=data["Sales"].mean()
print("avg_sale:",avg_sale)

profit_margin=(total_profit/sales_revenue)*100
print("profit_margin:",profit_margin)
```

```python
#Which product generates the highest total profit?

# Sum profit for each sub-category and sort from highest to lowest
profit_by_subcategory = (
    data.groupby("Sub-Category", as_index=False)["Profit"]
    .sum()
    .sort_values("Profit", ascending=False)
)

# Print the highest-profit sub-category
top = profit_by_subcategory.iloc[0]
print(f"Highest-profit sub-category: {top['Sub-Category']}")
print(f"Total profit: ${top['Profit']:,.2f}")

# Plot total profit by sub-category
plt.figure(figsize=(10, 7))
sns.barplot(
    data=profit_by_subcategory,
    x="Profit",
    y="Sub-Category",
    color="steelblue"
)
plt.title("Total Profit by Sub-Category")
plt.xlabel("Total profit ($)")
plt.ylabel("Sub-category")
plt.tight_layout()
plt.show()
```

```python
#watching the top 15 cities by technology sales

technology_by_city = (
    data[data["Category"] == "Technology"].groupby("City", as_index=False)["Sales"].sum().sort_values("Sales", ascending=False)
)

display(technology_by_city.head(15))

top_cities = technology_by_city.head(15)

plt.figure(figsize=(10, 6))
sns.barplot(data=top_cities, x="Sales", y="City", color="steelblue")
plt.title("Top 15 Cities by Technology Sales")
plt.xlabel("Total sales ($)")
plt.ylabel("City")
plt.tight_layout()
plt.show()
```

```python
technology_by_city = (
    data[data["Category"] == "Furniture"].groupby("City", as_index=False)["Sales"].sum().sort_values("Sales", ascending=False)
)

top_10_cities = technology_by_city.head(10)

sns.barplot(data=top_10_cities, x="Sales", y="City", color="steelblue")
plt.xlabel("Total sales ($)")
plt.ylabel("City")
plt.title("Top 10 Cities by Furniture Sales")
plt.tight_layout()
plt.show()
```
